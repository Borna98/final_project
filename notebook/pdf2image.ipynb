```python
from pathlib import Path
from pdf2image import convert_from_path

# 1. Ensure ROOT_DIR is a proper Path object (Wrapped in strings)
ROOT_DIR = Path("D:/DU_MIT/3rd_semester/DocumentIntelligence")

# 2. Correctly combine paths using strings for the folder names
DATASET_DIR = ROOT_DIR / "dataset"
OUTPUT_DIR = ROOT_DIR / "data"

# Fixed syntax errors in f-strings (added missing quotes)
print(f"🔍 Looking for PDFs in: {DATASET_DIR}")
print(f"📁 Images will be saved to: {OUTPUT_DIR}")
```

# InventoryMovements

```python
from pathlib import Path
from pdf2image import convert_from_path

dir1 = Path(ROOT_DIR / DATASET_DIR / "InventoryMovements")
# dir1 = Path(ROOT_DIR / DATASET_DIR / "Invoices")
# dir1 = Path(ROOT_DIR / DATASET_DIR / "PurchaseOrders")
# dir1 = Path(ROOT_DIR / DATASET_DIR / "ShippingOrders")
output_dir = dir1 / "images"
output_dir.mkdir(parents=True, exist_ok=True)  # Ensures the output directory exists

pdf_files = list(dir1.glob("*.pdf")) # [:10]

for pdf_path in pdf_files:
    print(f"\n⏳ Processing: {pdf_path.name}")

    # Convert the current PDF
    pages = convert_from_path(
        pdf_path, 
        dpi=300, 
        poppler_path="D:/DU_MIT/3rd_semester/DocumentIntelligence/library/poppler-23.05.0/Library/bin"
    )

    # Indented inside the main loop so it runs for EVERY PDF
    for i, page in enumerate(pages):
        # Added pdf_path.stem to include the original PDF name in the image filename
        output_filename = output_dir / f"{pdf_path.stem}_page_{i+1}.png"
        page.save(output_filename, "PNG")
        
    print(f"✅ Successfully saved {len(pages)} pages for {pdf_path.name}")
```
